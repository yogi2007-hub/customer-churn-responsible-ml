# Customer Churn: Majority-Class Baseline

This is a simple baseline for the 12-row training sample. It predicts the majority class (not churned) for every customer. The data provider's permission to publish customer-level rows is unknown, so this notebook uses only the class counts. These are training-sample results, not a test of real-world performance.

```python
# Target counts from the supplied 12-row sample: 5 churned, 7 not churned.
actual = [1] * 5 + [0] * 7
predicted = [0] * len(actual)  # majority-class baseline

tp = sum(a == 1 and p == 1 for a, p in zip(actual, predicted))
tn = sum(a == 0 and p == 0 for a, p in zip(actual, predicted))
fp = sum(a == 0 and p == 1 for a, p in zip(actual, predicted))
fn = sum(a == 1 and p == 0 for a, p in zip(actual, predicted))

accuracy = (tp + tn) / len(actual)
precision = tp / (tp + fp) if tp + fp else 0.0
recall = tp / (tp + fn) if tp + fn else 0.0
print(f"Confusion counts: TP={tp}, TN={tn}, FP={fp}, FN={fn}")
print(f"Accuracy:  {accuracy:.3f}")
print(f"Precision: {precision:.3f}")
print(f"Recall:    {recall:.3f}")
```

## Interpretation and limits

This baseline gets 7 of 12 rows correct (accuracy 0.583), but its churn recall is 0: it misses every churned customer. The sample is tiny, and these calculations use the same sample counts that define the baseline. They do not show how a model would perform on new customers. A larger, authorized dataset and a separate test set are needed before drawing conclusions.
